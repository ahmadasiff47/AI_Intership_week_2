```python
from datasets import load_dataset

# Config name 'en' add kar diya hai
dataset = load_dataset("FreedomIntelligence/medical-o1-reasoning-SFT", "en", split = "train[:1000]")

prompt_template = """Below is an instruction that describes a task, paired with an input that provides further context. Write a response that appropriately completes the request.

### Instruction:
{}

### Response:
{}"""

def format_prompts(examples):
    instructions = examples["Question"]
    outputs      = examples["Response"]
    texts = []
    for instruction, output in zip(instructions, outputs):
        text = prompt_template.format(instruction, output)
        texts.append(text)
    return { "text" : texts, }

dataset = dataset.map(format_prompts, batched = True)
print("Dataset successfully loaded and formatted!")
```

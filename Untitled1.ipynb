```python
%%writefile app.py
import os
import json
import time
import litellm
import gradio as gr
import plotly.graph_objects as go
import plotly.express as px
from duckduckgo_search import DDGS

# 🛠️ 1. Global Rate-Limit Guards & Retry Configurations
litellm.drop_params = True
litellm.num_retries = 10

try:
    import crewai.llms.cache as _crewai_cache
    _crewai_cache.mark_cache_breakpoint = lambda msg: msg
except (ImportError, AttributeError):
    pass

from crewai import Agent, Task, Crew, Process, LLM
from crewai.tools import tool

# 🌐 2. Search Intelligence Tool
@tool("Search Intelligence Tool")
def search_intelligence_tool(query: str) -> str:
    """Fetch live market context and competitive intelligence."""
    try:
        results = DDGS().text(query, max_results=2)
        return str(results)
    except Exception as e:
        return f"Search bypassed: {e}"

# 📊 3. Visual Analytics (Emerald & Gold Theme)
def render_financial_projection():
    fig = go.Figure()
    fig.add_trace(go.Bar(
        x=['Q1', 'Q2', 'Q3', 'Q4'],
        y=[200, 500, 1100, 2200],
        name='Projected Revenue ($K)',
        marker_color='#10b981'
    ))
    fig.add_trace(go.Scatter(
        x=['Q1', 'Q2', 'Q3', 'Q4'],
        y=[500, 1200, 3500, 8000],
        name='Active Enterprise Users',
        yaxis='y2',
        line=dict(color='#f59e0b', width=3, shape='spline')
    ))
    fig.update_layout(
        title="<b>📊 Revenue Trajectory & User Scaling</b>",
        template="plotly_dark",
        paper_bgcolor='rgba(6,78,59,0.1)',
        plot_bgcolor='rgba(15,23,42,0.4)',
        font=dict(family="Inter, sans-serif", color="#f8fafc"),
        yaxis=dict(title="Revenue ($K)", gridcolor="#1e293b"),
        yaxis2=dict(title="Active Users", overlaying='y', side='right'),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
    )
    return fig

def render_deployment_timeline():
    df = [
        dict(Sprint="Sprint 1: Technical Foundation", Start='2026-09-01', Finish='2026-09-20', Domain="Engineering"),
        dict(Sprint="Sprint 2: Security & Governance", Start='2026-09-15', Finish='2026-10-10', Domain="Compliance"),
        dict(Sprint="Sprint 3: GTM & Partnership Launch", Start='2026-10-05', Finish='2026-11-01', Domain="Commercial"),
        dict(Sprint="Sprint 4: Enterprise Scale", Start='2026-10-25', Finish='2026-11-25', Domain="Executive")
    ]
    fig = px.timeline(
        df, x_start="Start", x_end="Finish", y="Sprint", color="Domain",
        color_discrete_map={"Engineering": "#10b981", "Compliance": "#f59e0b", "Commercial": "#06b6d4", "Executive": "#a855f7"}
    )
    fig.update_layout(
        title="<b>🚀 Operational Deployment Roadmap</b>",
        template="plotly_dark",
        paper_bgcolor='rgba(6,78,59,0.1)',
        plot_bgcolor='rgba(15,23,42,0.4)',
        font=dict(family="Inter, sans-serif", color="#f8fafc")
    )
    return fig

# 🤖 4. Multi-Agent Venture Acceleration Engine
def run_venture_pulse(groq_key, venture_title, industry_domain, target_arr):
    chart_fin = render_financial_projection()
    chart_ops = render_deployment_timeline()

    if not groq_key or not groq_key.strip():
        return "⚠️ **Authentication Error:** Groq API Key is missing. Please enter your key in the control panel.", chart_fin, chart_ops, "❌ Key Required"

    os.environ["GROQ_API_KEY"] = groq_key.strip()

    try:
        # LLM setup using high-throughput model
        llm_engine = LLM(
            model="groq/llama-3.1-8b-instant",
            temperature=0.5,
            max_retries=10
        )

        # Specialized Agents
        scout = Agent(
            role="Market Intelligence Scout",
            goal=f"Analyze strategic opportunities for {venture_title} in {industry_domain}.",
            backstory="Specialist in identifying market gaps and competitive differentiators.",
            tools=[search_intelligence_tool],
            llm=llm_engine
        )

        architect = Agent(
            role="Technical Architecture Specialist",
            goal="Design scalable cloud infrastructure and LLM integration patterns.",
            backstory="Principal engineer focused on enterprise system security and reliability.",
            llm=llm_engine
        )

        monetization_lead = Agent(
            role="Revenue Model Strategist",
            goal=f"Build unit economics and pricing strategies targeting ${target_arr} ARR.",
            backstory="Financial strategist expert in SaaS monetization and margin optimization.",
            llm=llm_engine
        )

        growth_officer = Agent(
            role="Growth & Channel Officer",
            goal="Define enterprise acquisition channels and B2B positioning.",
            backstory="Marketing executive specializing in high-conversion GTM strategies.",
            tools=[search_intelligence_tool],
            llm=llm_engine
        )

        governance_lead = Agent(
            role="Risk & Governance Lead",
            goal="Outline compliance protocols, SLA standards, and risk mitigation.",
            backstory="Compliance officer ensuring enterprise readiness and operational safety.",
            llm=llm_engine
        )

        managing_director = Agent(
            role="Managing Director",
            goal="Synthesize all domain insights into an Executive Strategy Brief.",
            backstory="Executive leader aligning technical, financial, and market operations.",
            llm=llm_engine
        )

        # Context-Optimized Tasks
        t1 = Task(description=f"Identify key trends and competitors for {venture_title} in {industry_domain} in under 150 words.", expected_output="Market brief.", agent=scout)
        t2 = Task(description="Outline core system stack and security architecture in concise points.", expected_output="Tech architecture bullet points.", agent=architect, context=[t1])
        t3 = Task(description=f"Detail 3 tier pricing models to reach ${target_arr} ARR.", expected_output="Monetization model.", agent=monetization_lead, context=[t2])
        t4 = Task(description="Summarize 3 primary customer acquisition channels.", expected_output="GTM strategy points.", agent=growth_officer, context=[t1])
        t5 = Task(description="List 4 core operational milestones and risk mitigations.", expected_output="Governance & ops plan.", agent=governance_lead, context=[t2])
        t6 = Task(
            description="Synthesize team inputs into a comprehensive Executive Strategy Brief in clean Markdown.",
            expected_output="Final Strategy Brief.",
            agent=managing_director,
            context=[t3, t4, t5],
            output_file="Executive_Strategy_Brief.md"
        )

        swarm = Crew(
            agents=[scout, architect, monetization_lead, growth_officer, governance_lead, managing_director],
            tasks=[t1, t2, t3, t4, t5, t6],
            process=Process.sequential,
            max_rpm=2  # Prevents Groq Token-Per-Minute limit caps
        )

        result = swarm.kickoff()
        return str(result), chart_fin, chart_ops, "✅ Strategy Generation Complete"

    except Exception as e:
        return f"⚠️ **System Alert:** {str(e)}", chart_fin, chart_ops, f"❌ Status: {str(e)}"

# 🎨 5. UI Layout and Custom CSS
custom_theme = """
body, .gradio-container {
    background-color: #022c22 !important;
    color: #f8fafc !important;
    font-family: 'Inter', sans-serif !important;
}
.hero-header {
    background: linear-gradient(135deg, #064e3b 0%, #0f172a 100%);
    padding: 24px;
    border-radius: 14px;
    border: 1px solid #10b981;
    margin-bottom: 20px;
}
.action-btn {
    background: linear-gradient(90deg, #10b981 0%, #059669 100%) !important;
    border: none !important;
    color: white !important;
    font-size: 16px !important;
    font-weight: 700 !important;
    border-radius: 10px !important;
}
"""

with gr.Blocks(css=custom_theme, title="VenturePulse AI Accelerator") as demo:

    with gr.Column(elem_classes="hero-header"):
        gr.Markdown(
            """
            # ⚡ VenturePulse AI: Executive Growth Engine
            ### Multi-Agent Venture Acceleration Matrix Powered by **Groq Llama 3.1**
            """
        )

    with gr.Row():
        with gr.Column(scale=1):
            gr.Markdown("### ⚙️ Engine Controls")
            key_input = gr.Textbox(label="Groq API Key", placeholder="Enter your gsk_... key", type="password")
            name_input = gr.Textbox(label="Venture Name", value="VenturePulse AI")
            domain_input = gr.Textbox(label="Industry Domain", value="Enterprise AI & FinTech Analytics")
            arr_goal_input = gr.Textbox(label="Target ARR Goal", value="8M")

            start_btn = gr.Button("⚡ INITIATE STRATEGY MATRIX", elem_classes="action-btn")
            status_field = gr.Textbox(label="System Status", value="Awaiting user command...", interactive=False)

        with gr.Column(scale=2):
            with gr.Tabs():
                with gr.TabItem("📈 Scale & Delivery Analytics"):
                    gr.Markdown("#### Projected Financials & Roadmap")
                    plot_1 = gr.Plot(value=render_financial_projection())
                    plot_2 = gr.Plot(value=render_deployment_timeline())

                with gr.TabItem("📄 Executive Strategy Brief"):
                    brief_output = gr.Markdown("### Click 'INITIATE STRATEGY MATRIX' to generate your custom brief...")

    start_btn.click(
        fn=run_venture_pulse,
        inputs=[key_input, name_input, domain_input, arr_goal_input],
        outputs=[brief_output, plot_1, plot_2, status_field]
    )

if __name__ == '__main__':
    demo.launch(share=True, debug=True)
```
